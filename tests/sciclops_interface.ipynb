```python
# Calibration test loop

iterations = 10
plate_count = 14
source_stack = "stack1"
target_stack = "stack4"
for _ in range(iterations):
    for _ in range(plate_count):
        s.transfer_labware(
            source=source_stack,
            target="exchange",
            plate=Falcon_96_well,
        )

        s.remove_lid(
            source="exchange",
            target="lidnest1",
            plate=Falcon_96_well,
        )

        s.replace_lid(
            source="lidnest1",
            target="exchange",
            plate=Falcon_96_well,
        )

        s.transfer_labware(source="exchange", target=target_stack, plate=Falcon_96_well)
    source_stack, target_stack = target_stack, source_stack
```
